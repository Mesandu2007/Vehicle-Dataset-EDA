```python
# Outliers

Q1=df["price"].quantile(0.25)
Q3=df["price"].quantile(0.75)

IQR=Q3-Q1

lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
price_outliers=df[df["price"]>upper_bound]

print("Potential high-price outliers:", len(price_outliers))
```

```python
price_outliers.nlargest(15,"price")[
    ["price", "year", "manufacturer", "model", "odometer", "condition"] 
]
```

```python

odo_q1=df["odometer"].quantile(0.25)
odo_q3=df["odometer"].quantile(0.75)

odo_iqr = odo_q3 - odo_q1

odo_upper_bound=odo_q3 + 1.5*odo_iqr

print("Odometer Q1:", odo_q1)
print("Odometer Q3:", odo_q3)
print("Odometer IQR:", odo_iqr)
print("Upper bound:", odo_upper_bound)


```

```python
odometer_outliers=df[df["odometer"]> odo_upper_bound]

print("Potential high-mileage outliers:", len(odometer_outliers))

```
