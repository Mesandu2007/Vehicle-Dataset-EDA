```python
import matplotlib.pyplot as plt

years = [2022, 2023, 2024, 2025]
sales = [100, 150, 130, 220]

plt.plot(years, sales, marker="o")

plt.annotate(
    "Highest sales",
    xy=(2025, 220),
    xytext=(2023.5, 200),
    arrowprops=dict(arrowstyle="->")
)

plt.title("Annual Sales")
plt.xlabel("Year")
plt.ylabel("Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt

days=[1,2,3,4,5,6]

sales_a=[10,15,20,18,25,30]
sales_b=[8,12,17,22,24,28]



plt.plot(days,sales_a,color="blue",marker="o",label="Sales A")
plt.plot(days,sales_b,color="green",marker="s",label="Sales B")

plt.title("Daily Sales Comparison")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.xticks([1,2,3,4,5,6])
plt.ylim(0,35)

plt.legend()

plt.annotate(
    "Peak Sales",
    xy=(6,30),
    xytext=(4.5,33),
    arrowprops=dict(arrowstyle="->")



)
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
sales = [10, 20, 15, 25, 30]
profits = [5, 12, 8, 18, 22]

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# First chart
ax[0].plot(days, sales, marker="o")
ax[0].set_title("Daily Sales")
ax[0].set_xlabel("Day")
ax[0].set_ylabel("Sales")

# Second chart
ax[1].bar(days, profits, color="seagreen")
ax[1].set_title("Daily Profits")
ax[1].set_xlabel("Day")
ax[1].set_ylabel("Profit")

plt.tight_layout()
plt.show()
```

```python
# corelation heatmap

import numpy as np
import matplotlib.pyplot as plt

correlation = np.array([
    [1.00,  0.80, -0.60],
    [0.80,  1.00, -0.40],
    [-0.60, -0.40, 1.00]
])

labels = ["Price", "Year", "Mileage"]

plt.figure(figsize=(7, 5))

plt.imshow(correlation, cmap="coolwarm", vmin=-1, vmax=1)

plt.colorbar(label="Correlation")

plt.xticks(range(3), labels, rotation=45)
plt.yticks(range(3), labels)

plt.title("Correlation Heatmap")

for i in range(3):
    for j in range(3):
        plt.text(j, i, f"{correlation[i, j]:.2f}",
                 ha="center", va="center")

plt.tight_layout()
plt.show()
```
