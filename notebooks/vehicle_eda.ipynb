```python
# Price distribution

df["price"].describe()
```

```python
q1=df["price"].quantile(0.25)
q3=df["price"].quantile(0.75)

iqr=q3-q1

print("First quartile (Q1):", q1)
print("Third quartile (Q3):", q3)
print("Interquartile range (IQR):", iqr)
```

```python
#  outlier boundries
lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr
print("Lower boundary:", lower_bound)
print("Upper boundary:", upper_bound)
```

```python
high_price_outliers = df[df["price"] > upper_bound]

print("Number of potential high-price outliers:", len(high_price_outliers))

print(
    "Percentage of dataset:",
    round(len(high_price_outliers) / len(df) * 100, 2),
    "%"
)
```

```python
# Question 33 — Which manufacturers have the highest average listing prices?

df.groupby("manufacturer")["price"].mean().sort_values(ascending=False).head(10).round(2)


#**Finding:** Ferrari has the highest average listing price ($159,773.76), followed by Aston Martin ($109,944.00). This suggests that the listings for these manufacturers are positioned at the premium end of the dataset. Average prices may be influenced by specific models and small sample sizes.
```

```python
# Which manufacturers have the highest average mileage?

df.groupby("manufacturer")["odometer"].mean().sort_values(ascending=False).head(10).round(2)


#**Finding:** Pontiac has the highest average odometer reading among the manufacturers shown, at approximately 131,341.98 miles. Saturn and Mercury follow. These results describe the listings in this sample and may be influenced by vehicle age, model, and sample size.
```

```python
# Compare average prices among the most common manufacturers

top_manufacturers = df["manufacturer"].value_counts().head(10).index

df[df["manufacturer"].isin(top_manufacturers)].groupby("manufacturer")["price"].mean().sort_values(ascending=False).round(2)

# Finding: Among the 10 most common manufacturer labels, RAM has the highest average listing price at $31,479.62. Honda and Nissan have the lowest averages, at $12,911.10 and $12,907.95, respectively. These differences may be influenced by vehicle age, model, mileage, and condition.
```
